```python
x = df.drop(columns=['resale_price'])
```

```python
y = df['resale_price']
```

```python
col = ['brand' , 'model' ,'os_type' , 'city_tier' , 'seller_type','condition']

le = LabelEncoder()
for columns  in col :
    x[columns] =le.fit_transform(x[columns])
```

```python
y = le.fit_transform(y)
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y, test_size=0.2 ,random_state=42)
```

```python
scale = StandardScaler()
x_train = scale.fit_transform(x_train)
x_test = scale.transform(x_test)
```

```python
Model1 =RandomForestRegressor(random_state = 42 , n_estimators=100)
```

```python
Model1.fit(x_train, y_train)
```

```python
y_pred = Model1.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Calculate regression metrics
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Regression Evaluation Report")
print("-" * 30)
print(f"Mean Absolute Error (MAE) : {mae:.4f}")
print(f"Mean Squared Error (MSE)   : {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"R-squared Score (R²)      : {r2:.4f}")
```

```python
from xgboost import XGBRegressor
from sklearn.model_selection import train_test_split, RandomizedSearchCV
```

```python
xgb = XGBRegressor(
    objective='reg:squarederror', 
    random_state=42, 
    n_jobs=-1
)
```

```python
bg=xgb.fit(x_train, y_train)
y_pred = bg.predict(x_test)

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Regression Evaluation Report")
print("-" * 30)
print(f"Mean Absolute Error (MAE) : {mae:.4f}")
print(f"Mean Squared Error (MSE)   : {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"R-squared Score (R²)      : {r2:.4f}")
```

```python
param_distributions = {
    'n_estimators': [100, 200, 300, 500],
    'learning_rate': [0.01, 0.03, 0.05, 0.1, 0.2],
    'max_depth': [3, 5, 7, 9, 11],
    'subsample': [0.6, 0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.6, 0.7, 0.8, 0.9, 1.0],
}

xgb = XGBRegressor(
    objective='reg:squarederror', 
    random_state=42, 
    n_jobs=-1
)

xgb_random = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_distributions,
    n_iter=50,
    scoring='neg_mean_squared_error',
    cv=5,
    verbose=1,
    random_state=42,
    n_jobs=-1
)

# 8. Fit on training set
xgb_random.fit(x_train, y_train)

# 9. Evaluate Best Model
best_xgb = xgb_random.best_estimator_
y_pred = best_xgb.predict(x_test)

print("Best Parameters:", xgb_random.best_params_)
print("Test RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("Test R² Score:", r2_score(y_test, y_pred))
```
